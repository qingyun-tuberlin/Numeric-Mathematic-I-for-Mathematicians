```python
import numpy as np
```

## Matrix Operation

### det(A)

```python
A = np.array([[1,2],
              [3,4]])
 
np.linalg.det(A)
# 可以判断矩阵是否可逆. 如果结果 接近0，矩阵不可逆或数值不稳定。
```

### Ax = b

```python
# solve Ax = b
A = np.array([[1,2],
              [3,4]])
b = np.array([[1,0],[0,1]])
x = np.linalg.solve(A, b)
print(x)
```

### $A^{-1}$

```python
A = np.array([[1,2],
              [3,4]])
A_inv = np.linalg.inv(A)
print(A_inv)
```

### $\sum_{i=1}^{N}\sum_{j=1}^{N} A_{ij}$

```python
A = np.array([[1,2],
              [3,4]])
sum_A = np.sum(A)
print(sum_A)
```

### rowwise sum A

```python
A = np.array([[1,2],
              [3,4]])
row_sum_A = np.sum(A, axis=1)
print(row_sum_A)
```

### columwise sum A

```python
A = np.array([[1,2],
              [3,4]])
row_sum_A = np.sum(A, axis=0)
print(row_sum_A)
```
