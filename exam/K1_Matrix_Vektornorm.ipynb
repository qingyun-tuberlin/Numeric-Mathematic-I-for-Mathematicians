Schreibe Python3 Funktionen, die die folgenden Normen einer Matrix A ausgibt,  
wobei A als numpy ndarry gegeben ist.

```python
import numpy as np
```

$$
\|A\|_{\infty} =
\max_{i=1,\dots,M}
\left(
\sum_{j=1}^{N} |a_{ij}|
\right)
$$

```python
def zeilensummennorm(A: np.ndarray) -> float:
    return np.max(np.sum(np.abs(A),axis=1))
```

```python
A = np.array([[1,2,3],
              [4,5,6]])
print("dimention: ", A.shape)
print("axis = 0")
print(np.sum(A,axis = 0))
print("axis = 1")
print(np.sum(A,axis = 1))
```

$$
\|A\|_{1} =
\max_{j=1,\dots,N}
\left(
\sum_{i=1}^{M} |a_{ij}|
\right)
$$

```python
def spaltensummennorm(A:np.ndarray)->float:
    return np.max(np.sum(np.abs(A),axis=0))
```

$$
\|A\|_2 =
\left(\rho(A^*A)\right)^{1/2}
=
\left(\rho(\bar{A}^{T}A)\right)^{\frac{1}{2}}
$$

$$
\| A \|_{F} := 
\left(
(\sum_{i=1}^{M} \sum_{j=1}^{N} |a_{ij}|^2)^{\frac{1}{2}}
\right)
$$

```python
def frobeniusnorm(A:np.ndarray) -> float:
    return np.sqrt(np.sum((np.abs(A)**2)))
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

print(frobeniusnorm(A))

print(np.linalg.norm(A, 'fro'))
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

print(np.sum(A))
print(np.sum(A,axis=0))
print(np.sum(A,axis=1))
```
