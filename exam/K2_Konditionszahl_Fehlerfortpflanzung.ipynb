```python
import numpy as np
```

```python
def hilbert_matrix(N: int) -> np.ndarray:
    H = np.zeros((N, N), dtype=float)
    for i in range(N):
        for j in range(N):
            H[i, j] = 1.0 / (i + j + 1)
    return H
```

Konditionszahl

```python
N = 12
H = hilbert_matrix(N)

x_true = np.ones(N)
b = H @ x_true

x_computed = np.linalg.solve(H, b)

print("x_true:")
print(x_true)

print("\nx_computed:")
print(x_computed)

print("\nmax abs error:")
print(np.max(np.abs(x_computed - x_true)))

print("\ncondition number (2-norm):")
print(np.linalg.cond(H))
```

Fehlerfortpflanzung

```python
N = 12
H = hilbert_matrix(N)

x_true = np.ones(N)
b = H @ x_true

# tiny perturbation
b_perturbed = b.copy()
b_perturbed[-1] += 1e-10

x1 = np.linalg.solve(H, b)
x2 = np.linalg.solve(H, b_perturbed)

print("x1:")
print(x1)

print("\nx2:")
print(x2)

print("\nDifference between solutions:")
print(x2 - x1)

print("\nRelative change in b:")
print(np.linalg.norm(b_perturbed - b) / np.linalg.norm(b))

print("\nRelative change in x:")
print(np.linalg.norm(x2 - x1) / np.linalg.norm(x1))
```
